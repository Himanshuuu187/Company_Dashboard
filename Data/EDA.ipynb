```python
today = pd.Timestamp.today().normalize()
join = data['Joining Date']

total_months = (today.year - join.dt.year) * 12 + (today.month - join.dt.month)
total_months = total_months - (today.day < join.dt.day).astype(int)
total_months = total_months.clip(lower=0)

years = total_months // 12
months = total_months % 12

data['YearsOfService'] = years.astype(str) + ' years and ' + months.astype(str) + ' months'

print(data[['Employee ID', 'Joining Date', 'YearsOfService']].head(10))
```

```python
# Overwrites the original CSV with the current DataFrame (including YearsOfService).
# index=False stops pandas from writing an extra 0,1,2,... index column.
data.to_csv('employee_data_1000.csv', index=False)
```
